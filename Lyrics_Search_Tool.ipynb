## 2️⃣ Enter Your Genius API Token

Get your free token at: https://genius.com/api-clients

```python
import lyricsgenius as genius
from IPython.display import Markdown, display
import os

# Enter your Genius API token here
GENIUS_TOKEN = ""  # Paste your token between the quotes

# OR use Colab's secure input
if not GENIUS_TOKEN:
    from getpass import getpass
    GENIUS_TOKEN = getpass("Enter your Genius API token: ")

# Initialize the API
api = genius.Genius(GENIUS_TOKEN)
api.verbose = False  # Turn off status messages
api.remove_section_headers = True  # Clean up the lyrics

print("✓ API initialized successfully!")
```

## 3️⃣ Search for Lyrics

### Option A: Single Song Search

```python
# Enter the song details
artist_name = "The Animals"  # Change this
song_title = "House of the Rising Sun"  # Change this

print(f"\n🔍 Searching for: {song_title} by {artist_name}...\n")

try:
    song = api.search_song(song_title, artist_name)
    
    if song:
        print("="*80)
        print(f"🎵 {song.title}")
        print(f"👤 {song.artist}")
        print("="*80)
        print()
        print(song.lyrics)
        print()
        print("="*80)
        print(f"🔗 Full lyrics: {song.url}")
    else:
        print("❌ Song not found. Please check the spelling and try again.")
        
except Exception as e:
    print(f"❌ Error: {e}")
```

### Option B: Multiple Songs Search

Search for multiple songs at once!

```python
# List of songs to search (format: "Artist - Song Title")
songs_to_search = [
    "The Animals - House of the Rising Sun",
    "Rolê da Diagrama - Piseiro do Baião",
    # Add more songs here
]

print(f"\n🔍 Searching for {len(songs_to_search)} songs...\n")

results = []

for song_string in songs_to_search:
    try:
        # Parse the string
        if " - " in song_string:
            artist, title = song_string.split(" - ", 1)
            artist = artist.strip()
            title = title.strip()
        else:
            print(f"⚠️  Invalid format: {song_string}")
            continue
        
        print(f"Searching: {title} by {artist}...")
        
        song = api.search_song(title, artist)
        
        if song:
            results.append({
                'artist': song.artist,
                'title': song.title,
                'lyrics': song.lyrics,
                'url': song.url
            })
            print(f"  ✓ Found!\n")
        else:
            print(f"  ❌ Not found\n")
            
    except Exception as e:
        print(f"  ❌ Error: {e}\n")

print(f"\n{'='*80}")
print(f" Found {len(results)} / {len(songs_to_search)} songs ")
print(f"{'='*80}\n")
```

## 4️⃣ Display Results

View all the lyrics you found!

```python
# Display all found lyrics
for i, song in enumerate(results, 1):
    print("\n\n")
    print("#" * 80)
    print(f"## Song {i}/{len(results)}")
    print("#" * 80)
    print(f"\n🎵 **{song['title']}**")
    print(f"👤 **{song['artist']}**")
    print(f"🔗 {song['url']}\n")
    print("-" * 80)
    print(song['lyrics'])
    print("-" * 80)
```

## 5️⃣ Save Lyrics to Files

Save the lyrics to text files for later use.

```python
import os
import re

# Create output directory
output_dir = "downloaded_lyrics"
os.makedirs(output_dir, exist_ok=True)

def sanitize_filename(name):
    """Remove invalid characters from filename"""
    return re.sub(r'[<>:"/\\|?*]', '', name)

# Save each song to a file
for song in results:
    filename = f"{sanitize_filename(song['artist'])} - {sanitize_filename(song['title'])}.txt"
    filepath = os.path.join(output_dir, filename)
    
    with open(filepath, 'w', encoding='utf-8') as f:
        f.write(f"{song['title']}\n")
        f.write(f"by {song['artist']}\n")
        f.write(f"Source: {song['url']}\n")
        f.write("\n" + "="*60 + "\n\n")
        f.write(song['lyrics'])
    
    print(f"✓ Saved: {filename}")

print(f"\n✓ All lyrics saved to /{output_dir}/")
```

## 6️⃣ Download All Lyrics (Google Colab)

Download all lyrics as a ZIP file.

```python
import sys

IN_COLAB = 'google.colab' in sys.modules

if IN_COLAB:
    import shutil
    from google.colab import files
    
    # Create ZIP archive
    print("Creating ZIP archive...")
    shutil.make_archive('lyrics', 'zip', output_dir)
    
    # Download
    print("Starting download...")
    files.download('lyrics.zip')
    print("✓ Download complete!")
else:
    print(f"Not running in Colab. Files are saved in: {os.path.abspath(output_dir)}")
```

## 🎯 Interactive Search Widget

Use this interactive widget for quick searches!

```python
from ipywidgets import widgets, Layout
from IPython.display import display, clear_output

# Create input widgets
artist_input = widgets.Text(
    placeholder='Enter artist name',
    description='Artist:',
    layout=Layout(width='50%')
)

song_input = widgets.Text(
    placeholder='Enter song title',
    description='Song:',
    layout=Layout(width='50%')
)

search_button = widgets.Button(
    description='🔍 Search Lyrics',
    button_style='success',
    layout=Layout(width='200px')
)

output = widgets.Output()

def on_search_clicked(b):
    with output:
        clear_output()
        
        artist = artist_input.value.strip()
        title = song_input.value.strip()
        
        if not artist or not title:
            print("⚠️  Please enter both artist and song name")
            return
        
        print(f"🔍 Searching for: {title} by {artist}...\n")
        
        try:
            song = api.search_song(title, artist)
            
            if song:
                print("="*80)
                print(f"🎵 {song.title}")
                print(f"👤 {song.artist}")
                print("="*80)
                print()
                print(song.lyrics)
                print()
                print("="*80)
                print(f"🔗 {song.url}")
            else:
                print("❌ Song not found. Please check the spelling.")
        except Exception as e:
            print(f"❌ Error: {e}")

search_button.on_click(on_search_clicked)

# Display the widget
display(widgets.VBox([
    artist_input,
    song_input,
    search_button,
    output
]))
```

## 📝 Tips & Notes

### Common Issues:

1. **Song not found?**
   - Check spelling of artist and song
   - Try searching without special characters
   - Some songs may not be available on Genius

2. **API Rate Limits:**
   - Free tier allows reasonable usage
   - Add small delays between searches if doing many songs

3. **Alternative for Brazilian/Regional Songs:**
   - Genius works best for popular songs
   - For regional/niche songs, you may need specialized sources
